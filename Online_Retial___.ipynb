```python
df_sales
df_returns
customer_sales = (
    df_sales
    .dropna(subset=["Customer ID"])
    .groupby("Customer ID")
    .agg(
        Gross_Revenue=("Revenue", "sum"),
        Orders=("Invoice", "nunique"),
        Quantity=("Quantity", "sum"),
        First_Order=("InvoiceDate", "min"),
        Last_Order=("InvoiceDate", "max")
    )
    .reset_index()
)

customer_returns = (
    df_returns
    .dropna(subset=["Customer ID"])
    .groupby("Customer ID")
    .agg(
        Cancelled_Revenue=("CancelledRevenue", "sum"),
        Cancelled_Orders=("Invoice", "nunique")
    )
    .reset_index()
)

customer_value = customer_sales.merge(
    customer_returns,
    on="Customer ID",
    how="left"
)

customer_value[
    ["Cancelled_Revenue", "Cancelled_Orders"]
] = customer_value[
    ["Cancelled_Revenue", "Cancelled_Orders"]
].fillna(0)

customer_value["Net_Revenue"] = (
    customer_value["Gross_Revenue"]
    - customer_value["Cancelled_Revenue"]
)

customer_value["Return_Rate_%"] = (
    customer_value["Cancelled_Revenue"]
    / customer_value["Gross_Revenue"]
) * 100

customer_value.sort_values(
    "Net_Revenue",
    ascending=False
).head(15)

```

```python
snapshot_date = df_sales["InvoiceDate"].max() + pd.Timedelta(days=1)
snapshot_date
```

```python
rfm = customer_value.copy()

rfm["Recency"] = (
    snapshot_date - rfm["Last_Order"]
).dt.days

rfm["Frequency"] = rfm["Orders"]

rfm["Monetary"] = rfm["Net_Revenue"]
```

```python
rfm = rfm[rfm["Monetary"] > 0].copy()
```

```python
rfm[
    ["Customer ID", "Recency", "Frequency", "Monetary"]
].describe()
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    4,
    labels=[4, 3, 2, 1]
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    4,
    labels=[1, 2, 3, 4]
)
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)
```

```python
rfm = rfm[rfm["Monetary"] > 0.01].copy()
```

```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    4,
    labels=[4, 3, 2, 1]
)

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    4,
    labels=[1, 2, 3, 4]
)

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    4,
    labels=[1, 2, 3, 4]
)
```

```python
rfm["R_Score"] = rfm["R_Score"].astype(int)
rfm["F_Score"] = rfm["F_Score"].astype(int)
rfm["M_Score"] = rfm["M_Score"].astype(int)
```

```python
def segment_customer(row):
    R = row["R_Score"]
    F = row["F_Score"]
    M = row["M_Score"]

    if R >= 3 and F >= 3 and M >= 3:
        return "Champions"

    elif R >= 3 and F >= 3:
        return "Loyal Customers"

    elif R >= 3 and F <= 2:
        return "Potential Loyalists"

    elif R <= 2 and F >= 3 and M >= 3:
        return "At Risk"

    elif R == 1 and F <= 2:
        return "Lost Customers"

    else:
        return "Regular Customers"
```

```python
rfm["Segment"] = rfm.apply(
    segment_customer,
    axis=1
)
```

```python
rfm["Segment"].value_counts()
```

```python
segment_summary = (
    rfm.groupby("Segment")
    .agg(
        Customers=("Customer ID", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean"),
        Total_Revenue=("Monetary", "sum")
    )
    .reset_index()
)

segment_summary
```
